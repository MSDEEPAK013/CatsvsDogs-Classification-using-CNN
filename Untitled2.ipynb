```python
import kagglehub


path = kagglehub.dataset_download("salader/dogsvscats")

print("Path to dataset files:", path)
```

```python
import os

print(os.listdir(path))
```

```python
import tensorflow as tf
from tensorflow import keras
from keras.layers import Dense,Flatten,Conv2D,MaxPooling2D
from keras.models import Sequential

```

```python
train = os.path.join(path, 'train')
```

```python
test = os.path.join(path, 'test')
```

```python
train1=keras.utils.image_dataset_from_directory(
    train,
    labels='inferred',
    label_mode='int',
    color_mode='rgb',
    batch_size=32,
    image_size=(256, 256),
    shuffle=True,
)

test1=keras.utils.image_dataset_from_directory(
    test,
    labels='inferred',
    label_mode='int',
    batch_size=32,
    image_size=(256,256),
    shuffle=True,
)
```

```python
def normalize(image,label):
  image=tf.cast(image/255.0,tf.float32)
  return image,label

train2=train1.map(normalize)
test2=test1.map(normalize)
```
